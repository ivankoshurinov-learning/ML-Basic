Задание 1-4 Линейная алгебра в pdf

```python

# Задание 5 Линейная алгебра
import numpy as np
import matplotlib.pyplot as plt

# Генерация матрицы
n = 100
np.random.seed(42)
A = np.random.randn(n, n)

# SVD разложение
U, s, Vt = np.linalg.svd(A, full_matrices=False)

# Аппроксимация для разных рангов
ranks = np.arange(2, n + 1, 2)
errors = []

for r in ranks:
    A_tilde = U[:, :r] @ np.diag(s[:r]) @ Vt[:r, :]
    E = np.sqrt(np.sum((A - A_tilde) ** 2))
    errors.append(E)

# График
plt.plot(ranks, errors, 'b-o', markersize=4, linewidth=1)
plt.xlabel('Ранг r')
plt.ylabel('Ошибка E(r)')
plt.title('График зависимости ошибки апроксимации матрицы от ранга')
plt.grid(True)
plt.show()
```
