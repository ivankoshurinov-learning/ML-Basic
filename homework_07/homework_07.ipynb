### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
# ВАШ КОД ЗДЕСЬ
my_arr = np.zeros(10)
my_arr[4] = 1
print(my_arr)
my_arr_2d = my_arr.reshape(2, 5)
print(my_arr_2d)
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
# ВАШ КОД ЗДЕСЬ
range_arr = np.arange(10, 50)
print(range_arr)
reversed_arr = np.flip(range_arr)
print(reversed_arr)
even_nums = reversed_arr[reversed_arr % 2 == 0]
print(even_nums)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
# ВАШ КОД ЗДЕСЬ
arr_3x3 = np.arange(9).reshape(3, 3)
print(arr_3x3)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
# ВАШ КОД ЗДЕСЬ
arr_4x3x2 = np.random.randint(0, 100, size=(4, 3, 2))
print(arr_4x3x2)
min_arr_4x3x2 = np.min(arr_4x3x2)
print(min_arr_4x3x2)
max_arr_4x3x2 = np.max(arr_4x3x2)
print(max_arr_4x3x2)
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
# ВАШ КОД ЗДЕСЬ
A = np.random.rand(6, 4)
B = np.random.rand(4, 3)
C = np.dot(A, B)
print(C)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
# ВАШ КОД ЗДЕСЬ
rand_7x7 = np.random.rand(7, 7)
print(rand_7x7)
mean_val = np.mean(rand_7x7)
print(mean_val)
std_val = np.std(rand_7x7)
print(std_val)
normalized_7x7 = (rand_7x7 - mean_val) / std_val
print(normalized_7x7)
```

### Импортируйте: pandas, matplotlib, seaborn

### Загрузите датасет Tips из набора датасетов seaborn

```python
# ВАШ КОД ЗДЕСЬ
my_data = sns.load_dataset('tips')
print(my_data)
```

```python
# ВАШ КОД ЗДЕСЬ
print(my_data['total_bill'].max())
```

### Найдите количество курящих людей

```python
# ВАШ КОД ЗДЕСЬ
print((my_data['smoker'] == 'Yes').sum())
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
# ВАШ КОД ЗДЕСЬ
avg_bill_by_day = my_data.groupby('day')['total_bill'].mean()
print(avg_bill_by_day)
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
# ВАШ КОД ЗДЕСЬ
median_total = my_data['total_bill'].median()
high_bills = my_data[my_data['total_bill'] > median_total]
avg_tip_by_sex_high = high_bills.groupby('sex')['tip'].mean()
print(avg_tip_by_sex_high)
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
# ВАШ КОД ЗДЕСЬ
my_data['smoker_binary'] = my_data['smoker'].map({'Yes': 1, 'No': 0})
print(my_data[['smoker', 'smoker_binary']].head())
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
# ВАШ КОД ЗДЕСЬ
plt.figure()
sns.histplot(my_data['total_bill'], bins=20, color='darkcyan', edgecolor='black')
plt.title('Гистограмма: Распределение суммы счета')
plt.xlabel('Сумма счета ($)')
plt.ylabel('Количество наблюдений')
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
# ВАШ КОД ЗДЕСЬ
plt.figure()
sns.scatterplot(data=my_data, x='total_bill', y='tip', color='coral', alpha=0.6)
plt.title('Зависимость чаевых от суммы счета')
plt.xlabel('Сумма счета ($)')
plt.ylabel('Чаевые ($)')
plt.show()
```

### Постройте pairplot

```python
# ВАШ КОД ЗДЕСЬ
sns.pairplot(my_data)
plt.suptitle('Анализ взаимосвязей признаков', y=1.02)
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
# ВАШ КОД ЗДЕСЬ
plt.figure()
sns.boxplot(data=my_data, x='day', y='total_bill', hue='day', palette='Set2', order=['Thur', 'Fri', 'Sat', 'Sun'], legend=False)
plt.title('Зависимость суммы счета от дня недели')
plt.xlabel('День недели')
plt.ylabel('Сумма счета ($)')
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
# ВАШ КОД ЗДЕСЬ
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(data=my_data[my_data['time'] == 'Lunch'], x='tip', bins=15, color='teal', ax=axes[0])
axes[0].set_title('Распределение чаевых (Lunch)')
sns.histplot(data=my_data[my_data['time'] == 'Dinner'], x='tip', bins=15, color='purple', ax=axes[1])
axes[1].set_title('Распределение чаевых (Dinner)')
fig.suptitle('Чаевые в обед и за ужином')
plt.tight_layout()
plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
# ВАШ КОД ЗДЕСЬ
fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharex=True, sharey=True)
sns.scatterplot(data=my_data[my_data['sex'] == 'Male'], x='total_bill', y='tip', hue='smoker', style='smoker', palette='Set1', s=60, ax=axes[0])
axes[0].set_title('Мужчины (Male)')
sns.scatterplot(data=my_data[my_data['sex'] == 'Female'], x='total_bill', y='tip', hue='smoker', style='smoker', palette='Set1', s=60, ax=axes[1])
axes[1].set_title('Женщины (Female)')
fig.suptitle('Чаевые в зависимости от пола и статуса курения')
for ax in axes:
    ax.set_xlabel('Сумма счета ($)')
    ax.set_ylabel('Чаевые ($)')
plt.tight_layout()
plt.show()
```
